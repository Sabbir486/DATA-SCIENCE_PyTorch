# PyTorch Neural Network (NN) Module

```python
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
```

# Create Model Class (Single Neural Network)

```python
class Model(nn.Module):
    def __init__(self, num_features):

        super(Model, self).__init__()
        self.linear = nn.Linear(num_features, 1)    # Created Single Neuron
        self.sigmoid = nn.Sigmoid() # Activation Function

    def forward(self, features):
        x_out = self.linear(features)
        x_out = self.sigmoid(x_out)
        return x_out
```

# Creating Datasets

```python
features = torch.rand(10, 5) # 10 samples, 5 features

# Create an instance of the Model
model = Model(num_features=5)

# Forward pass through the model
output = model(features)

# Print the output
print(output)
```
