# Creating a Tensor

```python
# usinng empty tensor

empty_tensor = torch.empty(2, 3)
print(empty_tensor)
```

```python
# check type of the tensor

print(type(empty_tensor))
```

```python
# using zeros tensor

zeros_tensor = torch.zeros(2, 3)
print(zeros_tensor)
```

```python
# using ones tensor

ones_tensor = torch.ones(2, 3)
print(ones_tensor)
```

```python
# using random tensor

random_tensor = torch.rand(2, 3)    # Always give values (0-1)
print(random_tensor)
```

```python
# manual seed for reproducibility --> Will give same random value for next 50 times

seed = 50
torch.manual_seed(seed)

# Generate a new random tensor with the same seed
new_random_tensor = torch.rand(2, 3)
print(new_random_tensor)
```

```python
# using tensor with specific values

specific_tensor = torch.tensor([[1, 2, 3], [4, 5, 6]])
print(specific_tensor)
```

```python
# Other ways to create tensor

# arange tensor
arange_tensor = torch.arange(0, 10, 2)  # Start, End, Step
print(arange_tensor)

# linspace tensor
linspace_tensor = torch.linspace(0, 10, steps=5)  # Start, End, Number of steps
print(linspace_tensor)

# eye tensor
eye_tensor = torch.eye(3)  # Identity matrix of size 3x3
print(eye_tensor)

# using full tensor
full_tensor = torch.full((2, 3), 7)  # Create a tensor of size 2x3 filled with the value 7
print(full_tensor)
```

# Tensor Shapes

```python
new_tensor = torch.tensor([[1, 2], [3, 4]])
print(new_tensor)
```

```python
new_tensor.shape
```

```python
# By using empty tensor, we want to create a new same shape tensor with specific values

empty_new_tensor = torch.empty_like(new_tensor)  
print(empty_new_tensor)
```

```python
zeroes_new_tensor = torch.zeros_like(new_tensor)
print(zeroes_new_tensor)
```

```python
ones_new_tensor = torch.ones_like(new_tensor)
print(ones_new_tensor)
```

```python
random_new_tensor = torch.rand_like(new_tensor, dtype=torch.float32)  # Create a random tensor with the same shape as new_tensor
print(random_new_tensor)
```

# Tensor Data Types

```python
new_tensor.dtype
```

```python
tensor_data_type = torch.tensor([[1, 2], [3, 4]], dtype=torch.float32)
print(tensor_data_type)
print(tensor_data_type.dtype)
```

```python
# using to

new_tensor.to(torch.float32)
```
