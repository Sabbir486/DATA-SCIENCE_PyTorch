# Mathematical Operations

```python
import torch
print(torch.__version__)
```

```python
# 1. Scalar Operation
seed = 20
torch.manual_seed(seed)

x = torch.rand(2, 2)
x
```

```python
# Addition
seed = 20
torch.manual_seed(seed)

y = torch.rand(2, 2)
y

z = x + y
z
```

```python
# Subtraction
z = x - y
z
```

```python
# Multiplication
z = x * y
z
```

```python
# Division
z = x / y
z
```

```python
# int division
z = x // y
z
```

```python
# mod
z = x % y
z
```

```python
# power
z = x ** y
z
```
