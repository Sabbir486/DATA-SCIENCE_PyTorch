# Mathematical Operations

```python
import torch
print(torch.__version__)
```

```python
# 1. Scalar Operation
seed = 20
torch.manual_seed(seed)

x = torch.rand(2, 2)
x
```

```python
# Addition
seed = 20
torch.manual_seed(seed)

y = torch.rand(2, 2)
y

z = x + y
z
```

```python
# Subtraction
z = x - y
z
```

```python
# Multiplication
z = x * y
z
```

```python
# Division
z = x / y
z
```

```python
# int division
z = x // y
z
```

```python
# mod
z = x % y
z
```

```python
# power
z = x ** y
z
```

```python
# abs

c = torch.tensor([-1, -2, 3, -4])
c.abs()
```

```python
# negative

c = torch.tensor([-1, -2, 3, -4])
c.neg()
```

```python
# round

d = torch.tensor([1.2, 1.5, 1.7, 2.0])
d.round()
```

```python
# ceil

e = torch.tensor([1.2, 1.5, 1.7, 2.0])
e.ceil()
```

```python
# floor

f = torch.tensor([1.2, 1.5, 1.7, 2.0])
f.floor()
```

```python
# clamp
g = torch.tensor([1.0, 2.0, 3.0, 4.0, 1.5, 5.5])
g.clamp(min=2.0, max=3.0)
```

# Reduction Operations

```python
h = torch.randint(size=(3, 3), low=0, high=10)
h
```

```python
# sum
torch.sum(h)
```

```python
torch.sum(h, dim=1)  # sum along the rows
```

```python
torch.sum(h, dim=0)  # sum along the columns
```

# mean

```python
torch.mean(h.float())
```

```python
# mean along the rows
torch.mean(h.float(), dim=1)

# mean along the columns
torch.mean(h.float(), dim=0)
```

# median

```python
torch.median(h.float())
```

```python
# max
torch.max(h.float())
```

```python
# product
torch.prod(h.float())
```

```python
# standard deviation --> Square root of variance
torch.std(h.float())
```

```python
# variance --> Measure of the spread of the data; equation for variance is --> [[sum of (x - mean)^2] / n]
torch.var(h.float())
```

```python
# argmax --> Give the index of the maximum value in the tensor
torch.argmax(h.float())
```

```python
# argmin --> Give the index of the minimum value in the tensor
torch.argmin(h.float())
```
