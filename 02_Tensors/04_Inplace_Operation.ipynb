# Inplace Operation

```python
import torch
```

```python
a = torch.rand(2, 3)
b = torch.rand(2, 3)

print(a)
print(b)
```

```python
a.add_(b) # Inplace operation, modifies the original tensor a
```

```python
a
```

```python
b.relu_() # Inplace operation, modifies the original tensor b
```

```python
b
```

```python
a
```
