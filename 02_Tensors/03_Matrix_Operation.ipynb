# Matrix Operation

```python
import torch
```

```python
a = torch.tensor([[1, 2, 3], [4, 5, 6]])
b = torch.tensor([[7,8], [9, 10], [11, 12]])

print(a)
print(b)
```

```python
# Matrix multiplication

c = torch.matmul(a, b)
print(c)
```

```python
# Dot product

d = torch.dot(a.flatten(), b.flatten())
print(d)
```

```python
# Transpose

e = torch.transpose(a, 0, 1)
print(e)
```

# Comparison Operations

```python
f = torch.tensor([[1, 2], [3, 4]])
g = torch.tensor([[5, 6], [7, 8]])
```

```python
# Greater than

f > g
```

```python
# Not equal to

f != g
```

# Special Functions

```python
k = torch.tensor([[1, 2, -1], [3, 4, 5]], dtype=torch.float32)
```

```python
torch.log(k)
```

```python
torch.exp(k)
```

```python
torch.sqrt(k)
```

```python
torch.sigmoid(k) # sigmoid --> 1/(1+e^-x) --> Output for an individual score (spam vs not spam)
```

```python
torch.softmax(k, dim=1) # softmax --> e^x/sum(e^x) --> Output for a set of scores (spam vs not spam vs not sure)
```

```python
torch.relu(k) # for (k < 0) --> 0, for (k >= 0) --> k
```
