# Vector Input --> AutoGrad

```python
import torch
```

```python
x = torch.tensor([1., 2., 3.], requires_grad=True)   # input tensor --> leaf
x
```

```python
y = (x ** 2).mean()   # output tensor --> root
y
```

```python
y.backward()
x.grad
```

# Clearing Grad

```python
x = torch.tensor(2.0, requires_grad=True)
x
```

```python
y = x ** 2
y
```

```python
y.backward()
x.grad
```

```python
x.grad.zero_()  # reset the gradient to zero
```

# Disabling Gradient Tracking --> When only Forward Pass needed and backward pass not needed (During Prediction times)

```python
# option -1

x.requires_grad_(False)  # disable gradient tracking
x
```

```python
# option -2

z = x.detach()  # create a new tensor that shares the same data but does not require gradients
z
```
