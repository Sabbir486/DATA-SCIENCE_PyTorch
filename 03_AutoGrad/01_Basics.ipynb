# AutoGrad --> Is a tool which completed all derivatives of a Neural Network automatically.

```python
import torch
```

# 1st Example

```python
x = torch.tensor(3.0, requires_grad=True)   # x --> input value, requires_grad=True means we want to calculate the derivative of y with respect to x
x
```

```python
y = x ** 2
y
```

```python
y.backward()    # derivative of y with respect to x is calculated and stored in x.grad
```

```python
x.grad  # value of dy/dx = 2*x = 6.0
```

# 2nd Example --> Using Chain Rule

```python
x = torch.tensor(3.0, requires_grad=True)
```

```python
y = x ** 2
```

```python
z = torch.sin(y)
z
```

```python
z.backward()  # derivative of z with respect to x is calculated and stored in x.grad --> dz/dx = dz/dy * dy/dx = cos(y) * 2*x = cos(9) * 6.0
```

```python
x.grad
```
